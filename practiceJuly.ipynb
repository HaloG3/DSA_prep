```python
## largest element in the list , smallest 
s = [1,2,34,5,6,4,2,5,6,3,0,0,-1]
# with method => max, min 
# using loop 
def lar(nums):
    j = float('-inf')
    for i in nums:
        
        if i>j: 
            i,j = j,i
    return j
        
print(lar(s))


# or
```

```python
## finding 2nd largest 

def seclar(nums):
    n = len(nums)
    if n <2:
        return -1 
    j =float('-inf')
    k = float('-inf')
    n = len(nums)
    for i  in range(n): 
        if nums[i]>j:
            k = j
            j= nums[i]
        elif nums[i]>k and nums[i]!=j:
            k = nums[i]
          
    return k 
        
print(seclar(s))
```
