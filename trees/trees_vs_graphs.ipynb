Trees and Graphs in Data Structures

- graph is not an inbuilt data structure in Python.
- Graphs are constructed using combinations of **dictionaries** (for adjacency lists) or **lists** (for edge lists)
Trees and graphs are fundamental **non-linear data structures** used to represent relationships between objects, but they differ significantly in structure and application. A **tree** is a **hierarchical** structure with a single **root node**, where each node (except the root) has exactly **one parent**, and there is **exactly one unique path** between any two nodes; it is **acyclic** (contains no loops) and typically has **n-1 edges** for **n** nodes. 

### while
 a **graph** is a **network model** consisting of vertices and edges that can be **cyclic**, **disconnected**, or contain **multiple paths** between nodes, with **no designated root** and **no fixed limit** on the number of edges.

Key distinctions :
- **parent-child relationships** (strict in trees, arbitrary in graphs),
-  **traversal complexity** (straightforward in trees vs. requiring cycle detection in graphs), 
-  **use cases**. Trees are ideal for **hierarchical data** like **file systems**, **XML/HTML DOM**, and **decision trees**,
-  while graphs excel at modeling **complex networks** such as **social networks**, **transportation systems**, and **dependency graphs**.

| Feature | Tree | Graph |
| :--- | :--- | :--- |
| **Structure** | Hierarchical, acyclic, single root | Non-hierarchical, can be cyclic or acyclic |
| **Path** | Exactly one unique path between nodes | Multiple paths allowed |
| **Root Node** | Has exactly one root | No root node |
| **Edges** | $n-1$ edges for $n$ nodes | No fixed relation; can have many edges |
| **Loops** | Not allowed | Allowed |
| **Traversal** | Preorder, Inorder, Postorder | DFS, BFS |
| **Applications** | File systems, organizational charts | Social networks, routing, maps |
