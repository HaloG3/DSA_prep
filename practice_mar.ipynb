```python
import math 
def jump(array, target):
    n = len (array)
    step = int (math.sqrt(n))
    prev = 0 

    while array[min(step, n) - 1]<target:
        prev = step 
        step += int(math.sqrt(n))
        if prev>= n:
            return -1
        
    for i in range(prev, min(step,n)):
        if array[i] == target:

            return i 
    return -1
```

```python
def searchro(nums, target): 
    low = 0 
    high = len(nums)-1
    while low<= high :
        mid = (low+high)//2
        if nums[mid]==target:
            return mid
# if left half is rotated so the operation is in left half
        if nums[low]<=nums[mid]:
            if nums[low]<= target<nums[mid]:
                high = mid-1
            else: 
                low = mid+1

    # if right half is rotated then operation is in right half 
        else: 
            if nums[mid]<target<=nums[high]:
                low = mid+1
            else: 
                high = mid-1
    return -1 
```
