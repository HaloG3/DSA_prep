```python
# pattern 1 make a solid square of 4*4 
n = 4
for i in range(n):
    for j in range(n):
        print('*', end=' ')
    print()

```

```python
# 2. make a hollow square of 5*5
n = 5 
for i in range(n):
    for j in range(n):
        if i !=0 and i!= n-1 and 0<j<n-1:
            print(' ', end= ' ')
        else:
            print('*', end = ' ')
    print()
```

```python
# 3 make a triangle right angled left sided: 
n = 5 
for i in range(1,n+1):
    for j in range(i):
        print('*', end= ' ')
    print()
```

```python
# make a inverse tiangle: 
for i in range(1,n+1):
    for j in range(n+1-i):
        print('*', end= ' ')
    print()
```
