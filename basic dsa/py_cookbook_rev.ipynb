```python
# * expression = we can unpack too many values using * 

record = ('mohit', 'mohit@gmail.com', '9680', '78650','8484')

# now we can unpack the each element normally 
# but by using the * we can do this as 
name, email, *phone_no = record

print (phone_no) # all the digits can be unpacked by single variable


# we can use it as trialing such as 
*trailing, current = [1,2,4,5,6,4,3,2,2,10]
print(trailing, '=> trailing')
print(current,' =>current')

# we can unpack by combining with specific data type ops of any type of data form it as example of string: 
line = 'nobody:*:-2:-2:Unprivileged User:/var/empty:/usr/bin/false'
uname, *fields, bool_val= line.split(':')

print(fields, '=> fields ')
print(bool_val, '=> bool val of the line ')
```
