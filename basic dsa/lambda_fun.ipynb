## lambda function: 
- it is small one line anonymous function , that is defined without a name , and it looks like 
``` python 
lambda arguement : expression
```

- where lambda is keyword ,
-  arguement : take some arguement 
- expression: it will compute it to the arguement and return the results 



````python
add10 = lambda x : x+10

print(add10(20))
````

````python
points2d = [(1,10),(22,4),(13,4), (14,6)]
points2d_sorted= sorted(points2d)
print(points2d_sorted)

points2d_sor = sorted(points2d, key = lambda x: x[1])
print(points2d_sor)
````

````python
# map function with lambda function 
a = [9,4,3,2,1,0]
b = map(lambda x : x*2, a)
print(list(b))
````
