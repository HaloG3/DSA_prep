## 1. list: 
- mutable, changeable, iterable means it implementes the __getitem__() method and __iter__() method.
- we can use list in  real time or in on time . 
- it can contain multiple type of data , 
- memory structure: list memory is variable, we can append ,extend the list 
- internally they are dynamic arrays of pointers
- Each element in a list holds a reference to an object in memory, which adds overhead for object headers, 
- Lists store 8-byte pointers (on 64-bit systems) for each element, pointing to separate object locations scattered in memory, rather than storing contiguous values. 
- append, extend, etc type method are good use. 
#### operations of list 
- append() => add elemnt in the end of the list 
- extend(-) => it need a data structure to be added in list by fetch all its ememnt and then appending in the end of list 
- pop() => to pop the last element 
- remove => remove the first occurance of the element 
- we can find length of all iterables => len()
- in => checks element in list or note return boolean 
- insert any element at any position 
- we can slice the list 

    > lis[start:end:gap] => lis =[2, 3, 5, 6, 2, 3, 4]
                            lis[3:] => start from 3rd and upto last ;

- clear() => removes the all items form the list 
- sort() returns none but it will sort the list inplace
- copy() makes a shallow copy 

- List comprehensions provide a concise way to create lists
- we can use del statement to delete a index 
- del can be used for delete the entire variable 
### we can use lists as stack : 
- appending in last and poping from last is fast in

### we can use lists as queue : 
- lists are not efficient for this purpose.
-  While appends and pops from the end of list are fast, doing inserts or pops from the beginning of a list is slow (because all of the other elements have to be shifted by one).

# 3. sets 
- mutable, and unordered collections of unique, hashable elements. 
- unordered 
- can be initialesed by { }, and set() constructor 
-  Sets automatically remove duplicate values;
-  Implemented via hash tables, enabling efficient insertion, deletion, and lookup operations
- elements must be immutable (e.g., integers, strings, tuples), but the set itself can be modified.
    > explaination: Things in sets are stored using their value (the "hash" is a shortcut to this). References in Python are one-way only. A "value container" doesn't know which variables refer to it, so it cannot notify them of changes. So if you change change the value of something that is in a set, this likely changes its hash, but the set isn't notified of the change. And suddenly your set (or you dictionary keys) can have a duplicate, and this will end badly.
    A tuple that contains only hashable things is itself hashable. This is fairly frequent since the base types (numbers, strings, bytes) are hashable.
    So you can use such hashable tuples as set elements or as dictionary keys.
- The frozenset type is immutable and hashable — its contents cannot be altered after it is created; it can therefore be used as a dictionary key or as an element of another set.

### operation: 
- Use .add() for single items or .update() for multiple items; 
- use .remove() (raises error if missing) or .discard() (silent if missing) to delete elements.
- x in s: Test x for membership in s
- set.dijoint(sets other ) checks common element in sets: 
- set <= other : Test whether every element in the set is in other. Return boolean 
- set < other Test whether the set is a proper subset of other, that is, set <= other and set != other.
- set | other : Return a new set with elements from the set and all others
- set & other : return a new set with common element in all other
