### iterators and iterables 


**Iterables** are objects that can be looped over (e.g., lists, strings, tuples) because they implement the `__iter__()` method or `__getitem__()` method, which returns a fresh iterator every time.
>  **Iterators** are objects that implement both `__iter__()` and `__next__()` methods to return data one item at a time while maintaining the internal state of the iteration.

Key differences between the two include:

*   **Reusability**: Iterables can be iterated over multiple times, whereas iterators are exhausted after a single pass.
*   **State Management**: Iterators store the current position in the sequence, while iterables do not track iteration state.
*   **Functionality**: Only iterators support the `next()` function directly, while iterables must first be converted to an iterator using `iter()`.
*   **Memory Efficiency**: Iterators are generally more memory-efficient as they yield one element at a time, making them ideal for large datasets or infinite streams.

While every iterator is also an iterable (since it implements `__iter__()`), not every iterable is an iterator. For example, a list is an iterable but not an iterator; calling `iter()` on a list creates a separate `list_iterator` object that performs the actual iteration.


### product() computes the Cartesian product of input iterables

- it is memory-efficient

```python
# itertools: product , permutations, combinationsm accumulate, groupby ,and infinite iterators
from itertools import product
a = [2,3]
b = [4,3]

prod = product(a,b)
prod2 = product(a,b , repeat=2)
print(list(prod))
# The function returns an iterator, so you typically convert it to a list or iterate with a loop to view results
```

```python
c = [1,2]
d = [3]

prod2 = product(c,d , repeat=2)
print(list(prod2))
```

```python
# permutation 
from itertools import permutations
a = [1,4,5]

perm = permutations(a,2 ) # has some other input methods such as length = 2
print(list(perm))
```

```python
# combination 
from itertools import combinations 
a = [1,4,5]
comb = combinations(a, 2)
print(list(comb))

# it has a other method : combination with replacements in which it gives the iteself combination 
from itertools import combinations_with_replacement
comb_wr = combinations_with_replacement(a, 2)
print(list(comb_wr))

```

 ### Accumulate: default to add
 it yields a running total or cumulative sequence, making it ideal for generating running sums, products, or custom accumulation logic

```python
# accumulate  = 
from itertools import accumulate
import operator
a = [1,2,3,4,5,6]
acc = accumulate(a)
print(a)
print(list(acc))
# change default to multiplication 
numbers = [1, 2, 3, 4, 5]
cumulative_product = list(accumulate(numbers, operator.mul))
print(list(cumulative_product))
```

```python
# if we use other operator 
from itertools import combinations 
a = [1,4,5,3,5,6]
comb = accumulate(a, func= max) # return the maximum of the list till that point
print(list(comb))
```

```python
# group by  : that groups consecutive elements in an iterable based on a key function. 

from itertools import groupby  
a = []

a = [1,2,3,5,5]
gro = groupby(a , lambda small: small<3)
for key, value in gro:
    print(key, list(value))
```

### How It Works
The function takes an iterable and an optional key function (which defaults to the identity function if not provided).  It returns an iterator that yields pairs of (key, group), where the group is itself an iterator over the consecutive elements sharing that key.  Because the returned groups share the underlying iterable, the previous group is no longer visible once the groupby object advances; therefore, groups must be stored as lists if needed later. 

- Iterator Behavior: The group iterator is lazy and shares the source; you must convert it to a list immediately if you need to iterate over it multiple times or store it. 

```python
persons= [{'name': 'tim', 'age':25}, {'name': 'Dan', 'age':25}, {'name':'Lisa', 'age':27}, {'name':'Claire', 'age':28}]
```

```python
group_obj = groupby(persons, key = lambda x: x['age'])
for key, value in group_obj: 
    print(key, list(value))
```

```python
# infinite iterator 
from itertools import count, cycle, repeat
for i in count(10): # count infinitely 
    print(i)
    if i ==11: # use for stoping 
        break
```

```python
a = [1,2,3,4]
for i in cycle(a): # it will cycling in the loop again and agian 
    
    print(i)
    if i ==4:
        break 
```

```python
# repeat : it will repeat and we can give it a value util it repeat 
for i in repeat(a,5):
    print(i)
```

# how to sort in list in list by specific element =

```python
# given 
l = [('s', 1), ('h', 1), ('i', 3), ('l', 2), ('a', 2), ('j', 1), ('k', 1), ('t', 1)]
k = sorted(l, key = lambda x: x[1])
print(k)

# how to multiply the first element of tuple to the second element of tuple
ans= [char * count for char, count in k]
print(ans)
```
