## bfs : Breadth first search  for a graph : 
- it is a traversal algorithms , it starts from a node and traverse all the nodes level by level
- first it visit nodes directly adjacents to the root and then visits their adjacent nodes means level by level 
#### other popular algos : Dijkshtra ,prims, Kahn's algos are dependent on BFS
-  mainly use of bfs means where level order traversal is need we use BFS: 
    - for detecting cycle, 
    - find shortest path 



## brief explaination 
Breadth-First Search (BFS) is a graph traversal algorithm that explores nodes layer by layer, starting from a source node and visiting all its immediate neighbors before moving to the next level.  It uses a queue data structure to manage the order of exploration and is particularly effective for finding the shortest path in unweighted graphs.



```python
# lets buid a graph  from adjacency list 
# adj[][] = [[1, 2], [0, 2], [0, 1, 3, 4], [2], [2]]


def addEdge(adj , u,v):
    adj[u].append(v)
    adj[v].append(u)

v = 5
adj = [[] for _ in range(v)]

     
addEdge(adj, 1, 2)
addEdge(adj, 1, 0)
addEdge(adj, 2, 0)
addEdge(adj, 2, 3)
addEdge(adj, 2, 4)
print('Adjacency list is : ',adj)

# now do the bfs
# this is for simple graph means no connected components 
# we need a queue; we add the nodes in queue if we encounter a new node 
# a set = visited for  adding the unique nodes
# a list for storing the final result  
from collections import deque
def bfs(adj):
    V = len(adj)
    visited = [False] * V
    res = []
    
    src = 0
    q = deque()
    visited[src] = True
    q.append(src)

    while q:
        curr = q.popleft()
        res.append(curr)

        # visit all the unvisited
        # neighbours of current node
        for x in adj[curr]:
            if not visited[x]:
                visited[x] = True
                q.append(x)
                
    return res
print('bfs traversal is : ', bfs(adj))
    

```

```python
# when there are connected components 
from collections import deque

# BFS for a single connected component
def bfsConnected(adj, src, visited, res):
    q = deque()
    visited[src] = True
    q.append(src)

    while q:
        curr = q.popleft()
        res.append(curr)

        # visit all the unvisited
        # neighbours of current node
        for x in adj[curr]:
            if not visited[x]:
                visited[x] = True
                q.append(x)

# BFS for all components (handles disconnected graphs)
def bfs(adj):
    V = len(adj)
    visited = [False] * V
    res = []

    for i in range(V):
        if not visited[i]:
            bfsConnected(adj, i, visited, res)
    return res


def addEdge(adj, u, v):
    adj[u].append(v)
    adj[v].append(u)
  
  
if __name__ == "__main__":
    V = 6
    adj = []
    
    # creating adjacency list
    for i in range(V):
        adj.append([])
        
    addEdge(adj, 1, 2)
    addEdge(adj, 2, 0)
    addEdge(adj, 0, 3)
    addEdge(adj, 4, 5)

    res = bfs(adj)

    for node in res:
        print(node, end=" ")
```
